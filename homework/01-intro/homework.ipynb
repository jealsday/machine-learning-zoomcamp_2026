## Q6. Median value of horsepower

1. Median of `horsepower`
2. Most frequent value (mode) of `horsepower`
3. Fill missing values with the **mode**
4. Median again _ has it changed?

```python
median_before = df["horsepower"].median()
median_before
```

```python
most_frequent = df["horsepower"].mode()[0]
most_frequent
```

```python
hp = df["horsepower"].fillna(most_frequent)
median_after = hp.median()
print("before:", median_before, "after:", median_after)


df["horsepower"] = hp
```

## Q7. Sum of weights


```python
df_asia = df[df["origin"] == "Asia"]
X = df_asia[["vehicle_weight", "model_year"]].head(7).values
X
```

```python
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
w
```

```python
w.sum()
```
