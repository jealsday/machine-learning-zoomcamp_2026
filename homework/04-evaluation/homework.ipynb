# Homework 4 - Evaluation (Lead Scoring 2026)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, KFold
from sklearn.metrics import roc_auc_score
from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LogisticRegression
```

## Data preparation

```python
df = pd.read_csv("../../data/course_lead_scoring_2026.csv")

categorical = ["lead_source", "industry", "employment_status", "location"]
numerical = ["lead_score", "number_of_courses_viewed", "interaction_count", "annual_income"]
target = "converted"

df[categorical] = df[categorical].fillna("NA")
df[numerical] = df[numerical].fillna(0.0)

df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=1
)

y_train = df_train[target].values
y_val = df_val[target].values
```

## Question 1. ROC AUC feature importance

```python
aucs = {}
for col in numerical:
    auc = roc_auc_score(y_train, df_train[col])
    if auc < 0.5:
        auc = roc_auc_score(y_train, -df_train[col])
    aucs[col] = auc

pd.Series(aucs).sort_values(ascending=False)
```

## Question 2. Training the model

```python
features = categorical + numerical


def train(df_train, y_train, C=1.0):
    dv = DictVectorizer(sparse=False)
    X_train = dv.fit_transform(df_train[features].to_dict(orient="records"))

    model = LogisticRegression(solver="liblinear", C=C, max_iter=1000)
    model.fit(X_train, y_train)
    return dv, model


def predict(df, dv, model):
    X = dv.transform(df[features].to_dict(orient="records"))
    return model.predict_proba(X)[:, 1]


dv, model = train(df_train, y_train)
y_pred = predict(df_val, dv, model)
round(roc_auc_score(y_val, y_pred), 3)
```

## Question 3. Precision and recall

```python
rows = []
for t in np.linspace(0, 1, 101):
    predict_positive = y_pred >= t
    tp = (predict_positive & (y_val == 1)).sum()
    fp = (predict_positive & (y_val == 0)).sum()
    fn = (~predict_positive & (y_val == 1)).sum()

    precision = tp / (tp + fp) if tp + fp > 0 else 0.0
    recall = tp / (tp + fn) if tp + fn > 0 else 0.0
    rows.append((round(t, 2), precision, recall))

df_scores = pd.DataFrame(rows, columns=["threshold", "precision", "recall"])

plt.plot(df_scores.threshold, df_scores.precision, label="precision")
plt.plot(df_scores.threshold, df_scores.recall, label="recall")
plt.xlabel("threshold")
plt.legend()
plt.show()

valid = df_scores[(df_scores.precision > 0) | (df_scores.recall > 0)]
valid.loc[(valid.precision - valid.recall).abs().idxmin()]
```

## Question 4. F1 score

```python
p, r = df_scores.precision, df_scores.recall
df_scores["f1"] = np.where(p + r > 0, 2 * p * r / (p + r), 0.0)

df_scores.loc[df_scores.f1.idxmax()]
```

## Question 5. 5-Fold CV

```python
def cross_validate(C):
    kfold = KFold(n_splits=5, shuffle=True, random_state=1)
    scores = []
    for train_idx, val_idx in kfold.split(df_full_train):
        df_tr = df_full_train.iloc[train_idx]
        df_va = df_full_train.iloc[val_idx]

        dv, model = train(df_tr, df_tr[target].values, C=C)
        scores.append(roc_auc_score(df_va[target].values, predict(df_va, dv, model)))
    return scores


scores = cross_validate(C=1.0)
print(np.round(scores, 3))
round(np.std(scores), 3)
```

## Question 6. Hyperparameter tuning

```python
results = []
for C in [0.000001, 0.001, 1]:
    scores = cross_validate(C)
    mean, std = round(np.mean(scores), 3), round(np.std(scores), 3)
    results.append((C, mean, std))
    print(f"C={C}: {mean} +- {std}")

min(results, key=lambda x: (-x[1], x[2], x[0]))[0]
```
